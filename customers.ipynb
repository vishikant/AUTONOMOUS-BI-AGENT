```python

import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)

# Number of products to generate
n_products = 5000

# Category -> Subcategory -> (minimum price, maximum price)
categories = {
    "Electronics": {
        "Mobile": (5000, 150000),
        "Laptop": (25000, 250000),
        "Headphones": (500, 35000)
    },
    "Fashion": {
        "Men Clothing": (300, 12000),
        "Women Clothing": (300, 15000),
        "Footwear": (500, 20000)
    },
    "Home": {
        "Kitchen": (100, 25000),
        "Furniture": (1000, 200000),
        "Decor": (100, 30000)
    },
    "Beauty": {
        "Skincare": (100, 8000),
        "Makeup": (100, 10000),
        "Haircare": (100, 5000)
    },
    "Sports": {
        "Fitness": (200, 100000),
        "Outdoor": (300, 80000),
        "Sportswear": (300, 15000)
    }
}

# Realistic product name options
product_names = {
    "Mobile": ["Smartphone", "Android Phone", "5G Smartphone",
               "Foldable Phone", "Budget Smartphone"],
    "Laptop": ["Business Laptop", "Gaming Laptop",
               "Ultrabook", "Student Laptop", "Workstation Laptop"],
    "Headphones": ["Wireless Headphones", "Bluetooth Earbuds",
                   "Gaming Headset", "Wired Headphones"],
    "Men Clothing": ["Casual Shirt", "Jeans", "T-Shirt",
                     "Formal Trousers", "Jacket"],
    "Women Clothing": ["Kurti", "Dress", "Top",
                       "Ethnic Wear", "Women's Jacket"],
    "Footwear": ["Running Shoes", "Sneakers", "Formal Shoes",
                 "Sandals", "Sports Shoes"],
    "Kitchen": ["Cookware Set", "Mixer Grinder", "Water Bottle",
                "Pressure Cooker", "Kitchen Storage Set"],
    "Furniture": ["Office Chair", "Study Table", "Sofa",
                  "Bookshelf", "Dining Table"],
    "Decor": ["Wall Art", "Table Lamp", "Vase",
              "Wall Clock", "Decorative Mirror"],
    "Skincare": ["Face Moisturizer", "Sunscreen",
                 "Face Serum", "Cleanser", "Face Wash"],
    "Makeup": ["Lipstick", "Foundation", "Mascara",
               "Eyeshadow Palette", "Compact Powder"],
    "Haircare": ["Shampoo", "Conditioner", "Hair Serum",
                 "Hair Mask", "Hair Oil"],
    "Fitness": ["Dumbbell Set", "Treadmill", "Yoga Mat",
                "Exercise Cycle", "Resistance Bands"],
    "Outdoor": ["Camping Tent", "Hiking Backpack",
                "Sleeping Bag", "Trekking Poles", "Camping Chair"],
    "Sportswear": ["Sports T-Shirt", "Track Pants",
                   "Gym Leggings", "Sports Shorts", "Tracksuit"]
}

brands = {
    "Electronics": ["Samsung", "Apple", "Dell", "HP", "Lenovo",
                    "Sony", "boAt", "JBL", "OnePlus"],
    "Fashion": ["Nike", "Adidas", "Puma", "Levis", "Zara",
                "H&M", "Roadster", "Biba"],
    "Home": ["IKEA", "Prestige", "Philips", "Havells",
             "Home Centre", "Cello", "Wakefit"],
    "Beauty": ["Lakme", "Maybelline", "Nivea", "Minimalist",
               "Mamaearth", "L'Oreal", "Dove"],
    "Sports": ["Decathlon", "Nike", "Adidas", "Puma",
               "Yonex", "Cosco", "Cult"]
}

rows = []

# Flatten all valid category-subcategory combinations
subcategory_options = [
    (category, subcategory, price_range)
    for category, subcategories in categories.items()
    for subcategory, price_range in subcategories.items()
]

for i in range(1, n_products + 1):

    # Select a valid category and subcategory
    category, subcategory, price_range = (
        subcategory_options[
            np.random.randint(len(subcategory_options))
        ]
    )

    min_price, max_price = price_range

    # Use a log-uniform distribution:
    # lower and mid-range products are more common than premium ones
    price = round(
        np.exp(
            np.random.uniform(
                np.log(min_price),
                np.log(max_price)
            )
        )
    )

    # Estimate product cost as a proportion of selling price
    cost_percentage = np.random.uniform(0.55, 0.80)
    cost = round(price * cost_percentage, 2)

    # Select a product name and brand
    base_name = np.random.choice(product_names[subcategory])
    brand = np.random.choice(brands[category])

    product_name = f"{brand} {base_name}"

    rows.append({
        "product_id": f"P{i:06d}",
        "product_name": product_name,
        "category": category,
        "subcategory": subcategory,
        "brand": brand,
        "price": price,
        "cost": cost
    })

# Create DataFrame
products = pd.DataFrame(rows)

# Save CSV
output_path = Path("data/raw/products.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)

products.to_csv(output_path, index=False)

print("Products generated successfully!")
print("Shape:", products.shape)
print("Saved to:", output_path)

print("\nSample products:")
print(products.head(10))

print("\nProducts by category:")
print(products["category"].value_counts())

print("\nPrice summary:")
print(products.groupby("category")["price"].describe().round(2))
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------

np.random.seed(42)

DATA_DIR = Path("data/raw")
DATA_DIR.mkdir(parents=True, exist_ok=True)

START_DATE = pd.Timestamp("2025-01-01")
END_DATE = pd.Timestamp("2026-12-31")

N_ORDERS = 200_000  # Change to 500_000 if required

# ------------------------------------------
# 2. LOAD CUSTOMERS
# ------------------------------------------

customers = pd.read_csv(DATA_DIR / "customers.csv")

customers["signup_date"] = pd.to_datetime(
    customers["signup_date"]
)

# Keep customers with valid IDs and signup dates
customers = customers.dropna(
    subset=["customer_id", "signup_date"]
).copy()

customers = customers[
    (customers["signup_date"] <= END_DATE)
].reset_index(drop=True)

if customers.empty:
    raise ValueError("No eligible customers found.")

# ------------------------------------------
# 3. SELECT CUSTOMERS FOR ORDERS
# ------------------------------------------

# Customers can place multiple orders.
# Some customers will place no orders.
customer_indices = np.random.randint(
    0, len(customers), size=N_ORDERS
)

selected_customers = customers.iloc[
    customer_indices
].reset_index(drop=True)

# ------------------------------------------
# 4. GENERATE ORDER DATES
# ------------------------------------------

# Generate a random date between the project start
# and end dates, then ensure it is not before signup.

total_days = (END_DATE - START_DATE).days

random_offsets = np.random.randint(
    0, total_days + 1, size=N_ORDERS
)

random_dates = (
    START_DATE
    + pd.to_timedelta(random_offsets, unit="D")
)

signup_dates = selected_customers["signup_date"]

# If the randomly selected date precedes signup,
# select a date between signup and the project end.
early_mask = random_dates < signup_dates

random_dates = pd.Series(random_dates)

remaining_days = (
    END_DATE - signup_dates
).dt.days.to_numpy()

replacement_offsets = np.random.randint(
    0,
    remaining_days[early_mask] + 1,
    size=early_mask.sum()
)

random_dates.loc[early_mask] = (
    signup_dates[early_mask].reset_index(drop=True)
    + pd.to_timedelta(replacement_offsets, unit="D")
).to_numpy()

random_dates.loc[early_mask] = (
    signup_dates[early_mask].reset_index(drop=True)
    + pd.to_timedelta(replacement_offsets, unit="D")
).to_numpy()

# ------------------------------------------
# 5. ADD SEASONAL ORDER PATTERNS
# ------------------------------------------
# Increase order activity during October and November.
# These months often include major Indian festive sales.

order_dates = pd.to_datetime(random_dates)

month = order_dates.month.to_numpy()

seasonal_boost = np.where(
    np.isin(month, [10, 11]),
    1.35,
    1.0
)

# Increase weekend order activity
weekend_boost = np.where(
    order_dates.dayofweek.to_numpy() >= 5,
    1.15,
    1.0
)

# ------------------------------------------
# 6. GENERATE ORDER STATUS
# ------------------------------------------

# Illustrative business probabilities.
# Final outcomes depend on the random seed.

order_status = np.random.choice(
    ["Delivered", "Shipped", "Processing", "Cancelled"],
    size=N_ORDERS,
    p=[0.78, 0.08, 0.07, 0.07]
)

# ------------------------------------------
# 7. GENERATE PAYMENT METHODS
# ------------------------------------------

payment_method = np.random.choice(
    [
        "UPI",
        "Credit Card",
        "Debit Card",
        "Net Banking",
        "Cash on Delivery",
        "Wallet"
    ],
    size=N_ORDERS,
    p=[0.35, 0.20, 0.15, 0.10, 0.12, 0.08]
)

# ------------------------------------------
# 8. GET CUSTOMER CITY
# ------------------------------------------

if "city" in selected_customers.columns:
    order_city = selected_customers["city"].to_numpy()
else:
    order_city = np.random.choice(
        ["Delhi", "Mumbai", "Bangalore", "Hyderabad",
         "Chennai", "Pune", "Jaipur", "Kolkata"],
        size=N_ORDERS
    )

# ------------------------------------------
# 9. CREATE ORDERS DATAFRAME
# ------------------------------------------

orders = pd.DataFrame({
    "order_id": [
        f"O{i:07d}" for i in range(1, N_ORDERS + 1)
    ],
    "customer_id": selected_customers["customer_id"],
    "order_date": order_dates,
    "order_status": order_status,
    "payment_method": payment_method,
    "city": order_city
})

# Sort by date for easier analysis
orders = orders.sort_values(
    "order_date"
).reset_index(drop=True)

# ------------------------------------------
# 10. SAVE CSV
# ------------------------------------------

output_file = DATA_DIR / "orders.csv"

orders.to_csv(output_file, index=False)

print("orders.csv generated successfully!")
print("Shape:", orders.shape)
print("Saved to:", output_file.resolve())

print("\nSample records:")
print(orders.head(10))

print("\nOrder status distribution:")
print(orders["order_status"].value_counts())

print("\nMonthly order counts:")
print(
    orders.groupby(
        orders["order_date"].dt.to_period("M")
    ).size().head(12)
)

print("\nDate range:")
print(orders["order_date"].min())
print(orders["order_date"].max())

print("\nUnique customers who placed orders:")
print(orders["customer_id"].nunique())
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

np.random.seed(42)

# --------------------------------------------------
# 1. FOLDERS AND DATE RANGE
# --------------------------------------------------

DATA_DIR = Path("data/raw")
DATA_DIR.mkdir(parents=True, exist_ok=True)

START_DATE = pd.Timestamp("2025-01-01")
END_DATE = pd.Timestamp("2026-12-31")

# Load existing data
orders = pd.read_csv(DATA_DIR / "orders.csv")
customers = pd.read_csv(DATA_DIR / "customers.csv")

# Convert date columns
orders["order_date"] = pd.to_datetime(orders["order_date"])
customers["signup_date"] = pd.to_datetime(customers["signup_date"])

# --------------------------------------------------
# 2. PAYMENTS.CSV
# --------------------------------------------------

# One payment per order in this version.
# Payment happens between order date and 3 days later.

payment_rows = []

for _, order in orders.iterrows():

    order_date = order["order_date"]
    payment_date = order_date + pd.Timedelta(
        days=int(np.random.randint(0, 4))
    )

    status = order.get("order_status", "Delivered")

    if status == "Cancelled":
        payment_status = np.random.choice(
            ["Failed", "Refunded", "Success"],
            p=[0.30, 0.20, 0.50]
        )
    else:
        payment_status = np.random.choice(
            ["Success", "Failed"],
            p=[0.96, 0.04]
        )

    payment_method = np.random.choice(
        ["UPI", "Credit Card", "Debit Card",
         "Net Banking", "Cash on Delivery", "Wallet"],
        p=[0.35, 0.20, 0.15, 0.10, 0.12, 0.08]
    )

    # Amount will be updated using order_items.csv if available.
    amount = round(np.random.uniform(300, 15000), 2)

    payment_rows.append({
        "payment_id": f"PAY{len(payment_rows) + 1:07d}",
        "order_id": order["order_id"],
        "payment_date": payment_date,
        "payment_method": payment_method,
        "amount": amount,
        "payment_status": payment_status
    })

payments = pd.DataFrame(payment_rows)

# If order items exist, calculate the actual order amount
items_path = DATA_DIR / "order_items.csv"

if items_path.exists():
    items = pd.read_csv(items_path)

    items["line_amount"] = (
        items["quantity"]
        * items["unit_price"]
        * (1 - items["discount"] / 100)
    )

    order_amounts = items.groupby("order_id")[
        "line_amount"
    ].sum()

    payments["amount"] = (
        payments["order_id"]
        .map(order_amounts)
        .fillna(payments["amount"])
        .round(2)
    )

payments.to_csv(DATA_DIR / "payments.csv", index=False)

print("payments.csv created:", payments.shape)

# --------------------------------------------------
# 3. RETURNS.CSV
# --------------------------------------------------

# Only delivered orders can be returned.
# Return occurs 1 to 30 days after the order date.
# This version allows up to one returned line per order.

delivered_orders = orders[
    orders["order_status"].eq("Delivered")
].copy()

return_order_mask = np.random.random(
    len(delivered_orders)
) < 0.08

selected_orders = delivered_orders.loc[
    return_order_mask
].copy()

returns_rows = []

if items_path.exists():
    # Select a real product purchased in the order.
    items = pd.read_csv(items_path)

    for _, order in selected_orders.iterrows():

        order_items = items[
            items["order_id"] == order["order_id"]
        ]

        if order_items.empty:
            continue

        item = order_items.sample(
            n=1,
            random_state=int(np.random.randint(0, 1_000_000))
        ).iloc[0]

        return_date = order["order_date"] + pd.Timedelta(
            days=int(np.random.randint(1, 31))
        )

        return_amount = (
            item["quantity"]
            * item["unit_price"]
            * (1 - item["discount"] / 100)
        )

        returns_rows.append({
            "return_id": f"R{len(returns_rows) + 1:07d}",
            "order_id": order["order_id"],
            "product_id": item["product_id"],
            "return_date": return_date,
            "return_reason": np.random.choice(
                ["Damaged", "Wrong Product", "Wrong Size",
                 "Not as Expected", "Defective", "Late Delivery"],
                p=[0.20, 0.15, 0.20, 0.20, 0.15, 0.10]
            ),
            "refund_amount": round(return_amount, 2)
        })

else:
    print(
        "WARNING: order_items.csv not found. "
        "Returns cannot be linked to purchased products."
    )

returns = pd.DataFrame(
    returns_rows,
    columns=[
        "return_id", "order_id", "product_id",
        "return_date", "return_reason", "refund_amount"
    ]
)

returns.to_csv(DATA_DIR / "returns.csv", index=False)

print("returns.csv created:", returns.shape)

# --------------------------------------------------
# 4. MARKETING_CAMPAIGNS.CSV
# --------------------------------------------------

campaign_templates = [
    ("Republic Day Sale", "Google Ads"),
    ("Valentine's Special", "Instagram"),
    ("Holi Sale", "Facebook"),
    ("Summer Sale", "Google Ads"),
    ("Monsoon Deals", "YouTube"),
    ("Independence Day Sale", "Instagram"),
    ("Festive Sale", "Google Ads"),
    ("Diwali Mega Sale", "Facebook"),
    ("Black Friday Deals", "Email"),
    ("Year End Sale", "Influencer"),
    ("Customer Retention", "Email"),
    ("New Customer Offer", "Instagram"),
    ("Weekend Flash Sale", "Google Ads"),
    ("Clearance Sale", "Facebook"),
    ("Premium Customer Campaign", "Email"),
]

campaign_rows = []

for i, (campaign_name, channel) in enumerate(
    campaign_templates, start=1
):

    # Choose a campaign start date within the project period.
    latest_start = END_DATE - pd.Timedelta(days=30)

    start_date = START_DATE + pd.Timedelta(
        days=int(
            np.random.randint(
                0, (latest_start - START_DATE).days + 1
            )
        )
    )

    duration_days = int(np.random.randint(3, 22))
    end_date = min(
        start_date + pd.Timedelta(days=duration_days),
        END_DATE
    )

    campaign_rows.append({
        "campaign_id": f"CAM{i:04d}",
        "campaign_name": campaign_name,
        "channel": channel,
        "start_date": start_date,
        "end_date": end_date,
        "spend": round(
            np.random.uniform(10000, 500000), 2
        )
    })

campaigns = pd.DataFrame(campaign_rows)

campaigns.to_csv(
    DATA_DIR / "marketing_campaigns.csv",
    index=False
)

print("marketing_campaigns.csv created:", campaigns.shape)

# --------------------------------------------------
# 5. WEBSITE_SESSIONS.CSV
# --------------------------------------------------

# Generate 200,000 sessions.
# Session date cannot precede customer signup date.

n_sessions = 200_000

customer_dates = customers[
    ["customer_id", "signup_date"]
].copy()

# Exclude customers whose signup date is outside the timeline.
customer_dates = customer_dates[
    (customer_dates["signup_date"] >= START_DATE)
    & (customer_dates["signup_date"] <= END_DATE)
].reset_index(drop=True)

if customer_dates.empty:
    raise ValueError(
        "No customers have signup dates within the project timeline."
    )

selected_idx = np.random.randint(
    0, len(customer_dates), size=n_sessions
)

sessions_customers = customer_dates.iloc[
    selected_idx
].reset_index(drop=True)

session_rows = []

for i, customer in sessions_customers.iterrows():

    signup_date = customer["signup_date"]

    available_days = (
        END_DATE - signup_date
    ).days

    offset = int(
        np.random.randint(0, available_days + 1)
    )

    session_date = signup_date + pd.Timedelta(
        days=offset
    )

    converted = int(np.random.random() < 0.04)

    session_rows.append({
        "session_id": f"S{i + 1:07d}",
        "customer_id": customer["customer_id"],
        "session_date": session_date,
        "device": np.random.choice(
            ["Mobile", "Desktop", "Tablet"],
            p=[0.65, 0.30, 0.05]
        ),
        "traffic_source": np.random.choice(
            ["Google", "Instagram", "Facebook",
             "Direct", "Email", "YouTube", "Referral"],
            p=[0.30, 0.20, 0.12, 0.15, 0.08, 0.08, 0.07]
        ),
        "pages_viewed": int(
            np.random.choice([1, 2, 3, 4, 5, 6, 8, 10])
        ),
        "session_duration": int(
            np.random.gamma(shape=2.5, scale=90)
        ),
        "converted": converted
    })

sessions = pd.DataFrame(session_rows)

sessions.to_csv(
    DATA_DIR / "website_sessions.csv",
    index=False
)

print("website_sessions.csv created:", sessions.shape)

print("\nAll four files generated successfully!")
print("\nOutput folder:", DATA_DIR.resolve())
```
